```python
'''1) Create numpy array and perform following operation > Convert 1D array to 2D array'''
import numpy as np

# Create a 1D array
arr_1d = np.array([1, 2, 3, 4, 5, 6])

print("1D Array:")
print(arr_1d)

# Convert 1D array to 2D array (2 rows, 3 columns)
arr_2d = arr_1d.reshape(2, 3)

print("\n2D Array:")
print(arr_2d)
```

```python
'''Print Array Attributes(Like shape, dimenssion, data type, itemsize) '''
import numpy as np

# Create a NumPy array
arr = np.array([[10, 20, 30],
                [40, 50, 60]])

# Print attributes
print("Array:")
print(arr)

print("\nShape:", arr.shape)
print("Dimensions:", arr.ndim)
print("Data Type:", arr.dtype)
print("Item Size:", arr.itemsize, "bytes")
```

```python
''' Reverse a 1D NumPy array'''
import numpy as np

arr = np.array([1, 2, 3, 4, 5])

reversed_arr = arr[::-1]

print("Original Array:", arr)
print("Reversed Array:", reversed_arr)
```

```python
'''Create a 4x4x3 array and extract value at its second set, first row and last column'''
import numpy as np

# Create a 4×4×3 array
arr = np.arange(1, 49).reshape(4, 4, 3)

print("Array:\n", arr)

# Extract value at second set, first row, last column
value = arr[1, 0, -1]

print("\nExtracted Value:", value)
```

```python
''' Create a 4x4 and Extract Odd Rows and Even Columns '''
import numpy as np

# Create a 4×4 array
arr = np.arange(1, 17).reshape(4, 4)

print("Original Array:")
print(arr)

# Extract odd rows and even columns
result = arr[::2, 1::2]

print("\nOdd Rows and Even Columns:")
print(result)
```

```python
'''Slice the first two rows and first two columns of econd set from a 4x4x3 array'''
import numpy as np

# Create a 4×4×3 array
arr = np.arange(1, 49).reshape(4, 4, 3)

# Slice: second set, first two rows, first two columns
result = arr[1, :2, :2]

print("Sliced Array:")
print(result)
```

```python
''' Replace all odd numbers in a NumPy array with -1 by itterating using for loop [[23, 56, 78, 93], [71, 82,13, 24]]'''
import numpy as np

# Create NumPy array
arr = np.array([[23, 56, 78, 93],
                [71, 82, 13, 24]])

# Replace odd numbers with -1 using for loop
for i in range(arr.shape[0]):
    for j in range(arr.shape[1]):
        if arr[i][j] % 2 != 0:
            arr[i][j] = -1

print(arr)
```

```python
'''Get the indices of non-zero elements in an array [1, 0, 2, 0, 3, 0, 4]'''
import numpy as np

arr = np.array([1, 0, 2, 0, 3, 0, 4])

# Get indices of non-zero elements
indices = np.nonzero(arr)

print("Indices of non-zero elements:", indices)
```

```python
''' Perform arithmetic operations on two NumPy arrays element-wise Add two NumPy arrays element by element.'''
import numpy as np

# Create two arrays
arr1 = np.array([10, 20, 30, 40])
arr2 = np.array([1, 2, 3, 4])

# Addition
add = arr1 + arr2

# Subtraction
sub = arr1 - arr2

# Multiplication
mul = arr1 * arr2

# Division
div = arr1 / arr2

print("Array 1:", arr1)
print("Array 2:", arr2)

print("\nAddition:", add)
print("Subtraction:", sub)
print("Multiplication:", mul)
print("Division:", div)
```

```python
'''Multiply two NumPy arrays element by element.'''
import numpy as np

# Create two arrays
arr1 = np.array([1, 2, 3, 4])
arr2 = np.array([5, 6, 7, 8])

# Element-wise multiplication
result = arr1 * arr2

print("Array 1:", arr1)
print("Array 2:", arr2)
print("Multiplication Result:", result)
```

```python
''' Write a code to compute the dot product of two NumPy arrays arr1 = [15, 20, 25] arr2 = [10,40,37]'''
import numpy as np

arr1 = np.array([15, 20, 25])
arr2 = np.array([10, 40, 37])

# Compute dot product
result = np.dot(arr1, arr2)

print("Dot Product:", result)
```
