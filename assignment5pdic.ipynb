```python
# Write a Python program to repeat a tuple three times using the * operator.
# Define the original tuple
original_tuple = (1, 2, 3)

# Repeat it 3 times using the * operator
repeated_tuple = original_tuple * 3

print("Repeated Tuple:", repeated_tuple)
# Output: (1, 2, 3, 1, 2, 3, 1, 2, 3)
```

```python
#Join Three Separate Tuples (+ operator)
t1 = (1, 2)
t2 = (3, 4)
t3 = (5, 6)

# Combine them using the + operator
combined_tuple = t1 + t2 + t3

print("Combined Tuple:", combined_tuple)
# Output: (1, 2, 3, 4, 5, 6)
```

```python
#Check if an Element Exists (in keyword)
fruits = ("apple", "banana", "cherry")
search_element = "banana"

# Check existence using 'in'
if search_element in fruits:
    print(f"Yes, '{search_element}' exists in the tuple.")
else:
    print(f"No, '{search_element}' does not exist.")
```

```python
#Total, Highest, and Lowest Value (Without built-in functions)
numbers = (10, 45, 2, 89, 23, 11)

# Initialize variables using the first element
total = 0
highest = numbers[0]
lowest = numbers[0]

# Loop through the tuple to calculate values
for num in numbers:
    total += num
    if num > highest:
        highest = num
    if num < lowest:
        lowest = num

print("Total:", total)
print("Highest Value:", highest)
print("Lowest Value:", lowest)
```

```python
#Filter a Tuple (Values greater than 10)
n = (3, 14, 7, 22, 9, 41, 18, 5)

# Filter using a tuple comprehension / generator expression
filtered_n = tuple(x for x in n if x > 10)

print("Filtered Tuple:", filtered_n)
# Output: (14, 22, 41, 18)
```

```python
#6 Count Elements in a Set (Without len())
s = {"cat", "dog", "bird", "fish"}
count = 0

# Loop through the set to manually count elements
for item in s:
    count += 1

print("Number of elements in the set:", count)
```

```python
#7) Combine Two Sets (All unique elements)
set_a = {1, 2, 3}
set_b = {3, 4, 5}

# Combine using the union() method or the | operator
combined_set = set_a.union(set_b)

print("Combined Set:", combined_set)
# Output: {1, 2, 3, 4, 5}
```

```python
#8) Find Common Elements in Two Sets
s1 = {1, 2, 3, 4}
s2 = {3, 4, 5, 6}

# Find common elements using intersection() or the & operator
common_elements = s1.intersection(s2)

print("Common Elements:", common_elements)
# Output: {3, 4}
```

```python
#Find Elements in Either Set (Symmetric Difference)
s1 = {1, 2, 3, 4}
s2 = {3, 4, 5, 6}

# Find elements in either s1 or s2, but not both
either_set = s1.symmetric_difference(s2)

print("Elements in either set (but not both):", either_set)
# Output: {1, 2, 5, 6}
```
